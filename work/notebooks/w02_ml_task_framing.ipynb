## Selected Lane

I selected **Lane 2: Refresh / Content Opportunity Scoring**.

### What type of ML problem is this?

This problem is primarily a **classification** task because the goal is to predict whether a content page is likely to be declining and therefore needs a content refresh.

The output can also be used as a **ranking (scoring)** problem. Instead of only predicting "declining" or "not declining", the model assigns a probability to each page. Pages with higher probabilities are ranked higher and reviewed first by the content team.

This framing supports practical business decisions by helping prioritize limited content resources.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

# 2. Target or Proxy

## Target

The target for this project is whether a content page is **declining**.

In the starter dataset, this is represented by the `trend_direction` column. For modeling, I use a binary target called `is_declining`, where:

- **1** = the page is declining (`trend_direction = "down"`)
- **0** = the page is not declining

This target allows the model to estimate the likelihood that a page needs a content refresh.

## Why this target?

The content team ultimately wants to know which pages should be reviewed first. Predicting whether a page is declining provides a practical signal for prioritizing refresh efforts.

```python
import pandas as pd

df = pd.read_csv("content_refresh_anonymized.csv")

# Create binary target
df["is_declining"] = (
    df["trend_direction"]
    .str.lower()
    .eq("down")
    .astype(int)
)

print("Target Distribution")
print(df["is_declining"].value_counts())

print("\nTarget Percentage")
print((df["is_declining"].value_counts(normalize=True) * 100).round(2))
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*


## Primary Success Metric

The primary evaluation metric for this project is **Precision@50**.

### Why Precision@50?

The goal is not to classify every page perfectly. Instead, the content team wants to identify the **highest-priority pages** for refresh.

Precision@50 measures how many of the top 50 recommended pages are actually declining. A higher Precision@50 means the model is sending the team to the right pages first, making better use of limited time and resources.

### Secondary Metrics

In addition to Precision@50, I may also monitor:

- Precision
- Recall
- F1-score

These metrics provide additional insight into the model's overall classification performance, while Precision@50 remains the most business-relevant metric.

```python
primary_metric = "Precision@50"

print(f"Primary Success Metric: {primary_metric}")
print("Reason: The goal is to rank the pages that should be reviewed first.")
```

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*



## Unit of Analysis

The unit of analysis is **one content page**.

Each row in the dataset represents a single content page and contains descriptive features such as:

- Search volume
- Impressions
- CTR
- Average position
- Word count
- Content age
- Days since last update

The model makes one prediction for each page, estimating whether it is likely to be declining and should be prioritized for a content refresh.

```python
# Create target column
df["is_declining"] = (
    df["trend_direction"]
    .str.lower()
    .eq("down")
    .astype(int)
)

# Columns used in our ML framing
lane_df = df[
    [
        "content_id",
        "search_volume",
        "impressions_90d",
        "avg_position",
        "ctr",
        "word_count",
        "content_age_days",
        "days_since_last_update",
        "is_declining",
    ]
]

print(f"Shape: {lane_df.shape}")

lane_df.head()
```

### Observation

Each row represents one content page.

The features describe the page's current performance and characteristics, while the target (`is_declining`) indicates whether the page is declining. This makes the dataset suitable for training a supervised machine learning model.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule may work well in simple situations, for example:

> "Refresh every page that has not been updated for 180 days."

However, real content performance depends on multiple factors working together, such as impressions, CTR, average position, page age, search volume, and update history.

Machine learning can learn complex relationships between these features instead of relying on one manually defined rule.

In addition, the model produces a probability score that allows pages to be ranked by priority, helping the content team focus on the pages that are most likely to benefit from a refresh.
