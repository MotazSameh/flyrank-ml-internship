# 1. My Lane (or Freestyle) and Why

## Selected Lane

I selected **Lane 2: Refresh / Content Opportunity Scoring**.

### Why this lane?

Many websites contain hundreds or thousands of pages, but not all of them need to be updated at the same time. The challenge is deciding which pages should be refreshed first to maximize the impact of the team's effort.

This lane focuses on identifying pages that are likely to be declining in performance and ranking them according to their refresh priority. Instead of manually reviewing every page, a machine learning model can help prioritize the pages that deserve immediate attention.

I chose this lane because it addresses a practical business problem, aligns with the starter pipeline provided by FlyRank, and provides an opportunity to build an interpretable machine learning solution that supports real content decisions.

# 2. The Question

## Research Question

**Which content pages should be prioritized for refresh because they are most likely to be declining in performance?**

### Decision

The decision is determining which pages should be reviewed and refreshed first.

### Action

The SEO or content team will prioritize the highest-ranked pages for content updates instead of reviewing every page manually.

### Cost of a Wrong Recommendation

If the model incorrectly recommends a healthy page, the team may waste time updating content that does not need attention.

If the model fails to identify a truly declining page, that page may continue losing traffic, clicks, and business value before anyone notices the problem.

### Why Machine Learning?

Many factors can influence whether a page is declining, including impressions, CTR, page age, content length, and update history. These relationships are difficult to capture with simple hand-written rules.

A machine learning model can learn patterns from historical data and rank pages according to their likelihood of needing a refresh, helping the team make more informed decisions.

# 3. Quick Look at the Data

The following statistics provide a quick overview of the starter dataset and support my choice of the **Refresh / Content Opportunity Scoring** lane.

Instead of assuming that pages need refreshing, I first explore the data to understand whether declining pages show measurable differences from other pages.

```python
import pandas as pd

df = pd.read_csv("content_refresh_anonymized.csv")

# Create binary label
df["is_declining"] = (df["trend_direction"].str.lower() == "down")

print("=" * 60)
print("Dataset Overview")
print("=" * 60)

print(f"Total pages: {len(df):,}")

declining_rate = df["is_declining"].mean() * 100
print(f"Declining pages: {declining_rate:.2f}%")

print("\nAverage Impressions (90 days)")
print(df.groupby("is_declining")["impressions_90d"].mean().round(2))

print("\nAverage Days Since Last Update")
print(df.groupby("is_declining")["days_since_last_update"].mean().round(2))
```

### Observations

From the starter dataset, about **54.21%** of the pages are classified as declining, indicating that content refresh is a common challenge rather than a rare event.

Declining pages receive fewer average impressions (**4,919**) compared to non-declining pages (**5,533**), suggesting that declining content tends to lose visibility over time.

Additionally, declining pages have not been updated for a longer period on average (**49.25 days**) than non-declining pages (**42.37 days**). This supports the idea that update history may be a useful signal when prioritizing pages for content refresh.

These observations suggest that predicting refresh opportunities is a meaningful machine learning problem rather than simply applying a fixed rule.
