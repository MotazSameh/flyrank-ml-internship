### Five features and when they are available

1. **gsc_impressions** — knowable at the decision moment because Search Console records the page's observed impressions before a refresh decision.

2. **gsc_clicks** — knowable at the decision moment because Search Console records observed clicks before the decision.

3. **gsc_avg_position** — knowable at the decision moment because the page's observed search position is available from Search Console.

4. **ga4_pageviews** — knowable at the decision moment when GA4 data is available for the client and page.

5. **ga4_sessions** — knowable at the decision moment when GA4 data is available for the client and page.


```python
# ============================================================
# Leakage Trap
# ============================================================

from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

# Use a small sample to keep memory usage low
sample_keys = (
    df_march[
        ["client_hash_id", "content_hash_id"]
    ]
    .drop_duplicates()
    .sample(n=5000, random_state=42)
)

# Keep only the sampled March rows
march_sample = df_march.merge(
    sample_keys,
    on=["client_hash_id", "content_hash_id"],
    how="inner"
)

# Keep only the sampled April rows
april_sample = df_april.merge(
    sample_keys,
    on=["client_hash_id", "content_hash_id"],
    how="inner"
)

print("March sample rows:", len(march_sample))
print("April matching rows:", len(april_sample))
```

```python
# Aggregate only the small sampled subset

march_perf = (
    march_sample
    .groupby(["client_hash_id", "content_hash_id"])["gsc_impressions"]
    .sum()
    .rename("march_impressions")
)

april_perf = (
    april_sample
    .groupby(["client_hash_id", "content_hash_id"])["gsc_impressions"]
    .sum()
    .rename("april_impressions")
)

outcome = (
    march_perf
    .to_frame()
    .join(april_perf, how="left")
    .fillna(0)
)

# Future outcome
outcome["future_decline"] = (
    outcome["april_impressions"] < outcome["march_impressions"]
).astype(int)

print("Sampled content items:", len(outcome))

print("\nFuture decline distribution:")
print(outcome["future_decline"].value_counts())
```

```python
import numpy as np
features = [
    "gsc_impressions",
    "gsc_clicks",
    "gsc_avg_position"
]

march_features = (
    march_sample
    .groupby(["client_hash_id", "content_hash_id"])[features]
    .mean()
)

honest_df = (
    march_features
    .join(outcome["future_decline"], how="inner")
    .dropna()
)

X_honest = honest_df[features].replace(
    [np.inf, -np.inf], np.nan
).fillna(0)

y = honest_df["future_decline"]

honest_tree = DecisionTreeClassifier(
    max_depth=2,
    random_state=42
)

honest_tree.fit(X_honest, y)

honest_pred = honest_tree.predict(X_honest)

honest_accuracy = accuracy_score(y, honest_pred)

print("Honest model accuracy:",
      round(honest_accuracy, 3))
```

```python
# ============================================================
# DELIBERATE LEAKAGE
# ============================================================

leaky_df = honest_df.join(
    outcome["april_impressions"],
    how="left"
)

leaky_features = features + ["april_impressions"]

X_leaky = leaky_df[leaky_features].replace(
    [np.inf, -np.inf], np.nan
).fillna(0)

leaky_tree = DecisionTreeClassifier(
    max_depth=2,
    random_state=42
)

leaky_tree.fit(X_leaky, y)

leaky_pred = leaky_tree.predict(X_leaky)

leaky_accuracy = accuracy_score(y, leaky_pred)

print("Honest model accuracy:",
      round(honest_accuracy, 3))

print("Leaky model accuracy:",
      round(leaky_accuracy, 3))
```

```python
# Remove the leaked feature.
# It must not be part of the final feature set.

final_features = features

print("Final features after removing leakage:")
print(final_features)
```

### Leakage trap

I used April impressions as a deliberate leaked feature because April is the future outcome period relative to the March decision point. The model was allowed to see information that would not have been available when the decision was made.

The higher score from the leaky model is therefore not evidence of better ML performance. It is evidence of target leakage.

I removed `april_impressions` and kept only March features that were observable at the decision moment.


## 4. Data limits

*What can this data never tell you? Unbalanced history, GSC-only early rows, window overlaps.*

### 4. Data limits

This data can support observed and directional analysis of content performance, but it has important limitations.

* **Uneven data availability:** In March 2026, GSC data was available for 36.69% of rows, while GA4 data was available for only 4.21% of rows. Therefore, GA4-based analysis may not represent the full dataset.

* **Limited time window:** This analysis uses March 2026 as a mid-panel development month. A single month cannot represent all seasonal or long-term changes in content performance.

* **No causal conclusions:** The data can show associations and directional patterns, but it cannot prove that a particular content change caused a performance change.

* **Future information must be excluded:** Features derived from a future outcome window cannot be used at the decision moment because they would create leakage.

* **Client-level variation:** Different clients may have different levels of data availability and different performance patterns, so results from one client or group of clients may not generalize to all clients.


```python
# Data availability limitation

total_rows = len(df_march)

gsc_available = df_march["gsc_data_available"].eq(True).sum()
ga4_available = df_march["ga4_data_available"].eq(True).sum()

print("Total rows:", total_rows)
print(f"GSC available: {gsc_available:,} ({gsc_available / total_rows * 100:.2f}%)")
print(f"GA4 available: {ga4_available:,} ({ga4_available / total_rows * 100:.2f}%)")

print("\nLimitation:")
print("GA4 coverage is limited, so GA4-based features may not represent all rows.")
```
